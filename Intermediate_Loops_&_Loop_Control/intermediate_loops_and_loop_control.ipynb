## 1. Print numbers from 1–100 skipping multiples of 5 using `continue`

**Problem Statement:**
Iterate from 1 to 100 and print each number, but skip all numbers that are divisible by 5 using the `continue` statement.

````python
print("Numbers from 1 to 100 (skipping multiples of 5):")

# Loop through 1 to 100
for num in range(1, 101):
    # Skip numbers divisible by 5
    if num % 5 == 0:
        continue
    print(num, end=" ")
print()
````

## 2. Stop at First Number Divisible by Both 7 and 11

**Problem Statement:**
Iterate from 1 to 100 and terminate the loop (`break`) as soon as the first number divisible by both 7 and 11 is encountered.

````python
print("Searching for the first number divisible by both 7 and 11 between 1 and 100...")

for num in range(1, 101):
    # Check divisibility by both 7 and 11
    if num % 7 == 0 and num % 11 == 0:
        print(f"Found: {num} (Divisible by 7 and 11: {num} / 7 = {num//7}, {num} / 11 = {num//11})")
        print(f"Loop terminated at number: {num}")
        break
````

## 3. Search for a Number in a List using `for-else`

**Problem Statement:**
Search for a user-provided number inside a list. Use Python's `for-else` construct to print:
- `"Number Found"` (and break)
- `"Number Not Found"` (in the `else` block if loop completes without break)

````python
def search_in_list(numbers, target):
    for item in numbers:
        if item == target:
            print("Number Found")
            break
    else:
        print("Number Not Found")

numbers_list = [14, 28, 35, 42, 59, 70, 88]
print(f"Target List: {numbers_list}")

print("\n--- Test 1: Searching for 42 ---")
search_in_list(numbers_list, 42)

print("\n--- Test 2: Searching for 99 ---")
search_in_list(numbers_list, 99)
````

## 4. Display Numbered Names using `enumerate()`

**Problem Statement:**
Given `names = ["Aman", "Ravi", "Sudhanshu", "Priya", "Anjali"]`, use `enumerate()` to display:
```
1 Aman
2 Ravi
3 Sudhanshu
4 Priya
5 Anjali
```

````python
names = ["Aman", "Ravi", "Sudhanshu", "Priya", "Anjali"]

# Using enumerate() with start index 1
for index, name in enumerate(names, start=1):
    print(f"{index} {name}")
````

## 5. Half-Pyramid Star Pattern

**Problem Statement:**
Print the following pattern:
```
*
**
***
****
*****
```

````python
rows = 5

# Outer loop for rows (1 to 5)
for i in range(1, rows + 1):
    # Inner loop for stars in each row
    for j in range(1, i + 1):
        print("*", end="")
    print()  # Newline after each row
````

## 6. Inverted Half-Pyramid Star Pattern

**Problem Statement:**
Print the following pattern:
```
*****
****
***
**
*
```

````python
rows = 5

# Outer loop counting down from 5 to 1
for i in range(rows, 0, -1):
    # Inner loop prints stars equal to current row count
    for j in range(1, i + 1):
        print("*", end="")
    print()  # Newline after each row
````

## 7. Generate Multiplication Tables from 1 to 10 using Nested Loops

**Problem Statement:**
Generate complete multiplication tables from 1 to 10 using nested loops.

````python
print("=" * 70)
print("                    MULTIPLICATION TABLES (1 to 10)                   ")
print("=" * 70)

# Outer loop for table numbers 1 to 10
for i in range(1, 11):
    print(f"\n--- Table of {i} ---")
    # Inner loop for multipliers 1 to 10
    for j in range(1, 11):
        print(f"{i} x {j:2d} = {i * j:3d}")
````

## 8. Numbers Between 1 and 200 Divisible by Both 3 and 5

**Problem Statement:**
Find all numbers between 1 and 200 that are divisible by both 3 and 5 (i.e. multiples of 15).

````python
divisible_numbers = []

# Loop through numbers from 1 to 200
for num in range(1, 201):
    # Check divisibility by both 3 and 5
    if num % 3 == 0 and num % 5 == 0:
        divisible_numbers.append(num)

print(f"Numbers between 1 and 200 divisible by both 3 and 5 (Total count: {len(divisible_numbers)}):")
print(divisible_numbers)
````

## 9. Remove Duplicates from a List (without using `set()`)

**Problem Statement:**
Given a list containing duplicate elements, create another list containing only unique elements **without using `set()`**.

````python
original_list = [10, 20, 10, 30, 40, 20, 50, 60, 30, 70, 80, 10]
unique_list = []

# Iterate through items and preserve only first appearances
for item in original_list:
    if item not in unique_list:
        unique_list.append(item)

print(f"Original List with Duplicates: {original_list}")
print(f"Unique Elements List         : {unique_list}")
print(f"Removed Duplicates Count     : {len(original_list) - len(unique_list)}")
````

## 10. Count Positives, Negatives, and Zeros

**Problem Statement:**
Given `numbers = [10, -4, 8, -2, 0, 15, -9, 21]`, count:
- **Positive numbers**
- **Negative numbers**
- **Zeros**

````python
numbers = [10, -4, 8, -2, 0, 15, -9, 21]

# Initialize counters
positive_count = 0
negative_count = 0
zero_count = 0

# Classify each number using conditional statements
for num in numbers:
    if num > 0:
        positive_count += 1
    elif num < 0:
        negative_count += 1
    else:
        zero_count += 1

print(f"Input Numbers: {numbers}")
print("=" * 36)
print("         COUNT SUMMARY              ")
print("=" * 36)
print(f"• Positive numbers (> 0) : {positive_count}")
print(f"• Negative numbers (< 0) : {negative_count}")
print(f"• Zeros (= 0)            : {zero_count}")
print(f"• Total elements         : {len(numbers)}")
print("=" * 36)
````

## 11. Prime Number Checker using a Loop

**Problem Statement:**
Write a program to determine whether a given number is prime using a loop.

````python
def is_prime(n):
    if n <= 1:
        return False, "<= 1 is not prime"
    
    # Check divisors from 2 up to sqrt(n)
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False, f"divisible by {i}"
    else:
        return True, "PRIME"

test_values = [1, 2, 17, 25, 29, 49, 97, 100]

print("=" * 40)
print("         PRIME CHECKER RESULTS          ")
print("=" * 40)
for val in test_values:
    prime_flag, reason = is_prime(val)
    status = "PRIME" if prime_flag else f"NOT Prime ({reason})"
    print(f"• {val:3d} -> {status}")
print("=" * 40)
````

## 12. Print All Prime Numbers Between 1 and 100

**Problem Statement:**
Print all prime numbers between 1 and 100 using loops.

````python
primes_1_to_100 = []

# Iterate from 1 to 100
for num in range(1, 101):
    if num > 1:
        # Test divisibility
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                break
        else:
            # Executed only if no divisor was found
            primes_1_to_100.append(num)

print(f"Prime Numbers between 1 and 100 (Total: {len(primes_1_to_100)} primes):")
print(primes_1_to_100)
````
