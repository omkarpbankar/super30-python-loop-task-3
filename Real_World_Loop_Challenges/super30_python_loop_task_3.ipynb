## 1. Calculate Total Transaction Value

**Problem Statement:**
Given `transactions = [1200, 450, 800, 1500, 2300, 700, 100]`, calculate the total transaction value **without using `sum()`**.

````python
# Given list of transactions
transactions = [1200, 450, 800, 1500, 2300, 700, 100]

# Initialize accumulator variable
total_value = 0

# Iterate through each transaction and compute running sum
for amount in transactions:
    total_value += amount

print(f"Transactions List: {transactions}")
print(f"Total Transaction Value: ₹{total_value}")
````

## 2. Highest and Lowest Transaction

**Problem Statement:**
From the same list `transactions = [1200, 450, 800, 1500, 2300, 700, 100]`, find the highest and lowest transaction **without using `max()` and `min()`**.

````python
transactions = [1200, 450, 800, 1500, 2300, 700, 100]

# Initialize highest and lowest with the first element
if transactions:
    highest = transactions[0]
    lowest = transactions[0]

    # Loop through the list to compare and update values
    for amount in transactions:
        if amount > highest:
            highest = amount
        if amount < lowest:
            lowest = amount

    print(f"Transactions: {transactions}")
    print(f"Highest Transaction: ₹{highest}")
    print(f"Lowest Transaction : ₹{lowest}")
else:
    print("Transaction list is empty.")
````

## 3. Average Temperature

**Problem Statement:**
Given `temperatures = [32, 35, 28, 40, 38, 31, 42]`, find the average temperature.

````python
temperatures = [32, 35, 28, 40, 38, 31, 42]

total_temp = 0
count = 0

# Calculate sum and count using a loop
for temp in temperatures:
    total_temp += temp
    count += 1

avg_temp = total_temp / count if count > 0 else 0

print(f"Recorded Temperatures: {temperatures}")
print(f"Sum of Temperatures: {total_temp}°C")
print(f"Total Readings: {count}")
print(f"Average Temperature: {avg_temp:.2f}°C")
````

## 4. Student Marks Categorization

**Problem Statement:**
Given student marks `marks = [78, 92, 45, 67, 88, 53, 99]`,
count how many students scored:
- **90+**
- **75–89**
- **50–74**
- **below 50**

````python
marks = [78, 92, 45, 67, 88, 53, 99]

# Initialize counters for each category
count_90_plus = 0
count_75_89 = 0
count_50_74 = 0
count_below_50 = 0

# Categorize marks using conditionals inside loop
for score in marks:
    if score >= 90:
        count_90_plus += 1
    elif score >= 75:     # 75 <= score <= 89
        count_75_89 += 1
    elif score >= 50:     # 50 <= score <= 74
        count_50_74 += 1
    else:                 # score < 50
        count_below_50 += 1

print(f"Student Marks: {marks}")
print("=" * 37)
print("        STUDENT SCORE REPORT         ")
print("=" * 37)
print(f"• 90+      : {count_90_plus} student(s)")
print(f"• 75 - 89  : {count_75_89} student(s)")
print(f"• 50 - 74  : {count_50_74} student(s)")
print(f"• Below 50 : {count_below_50} student(s)")
print("=" * 37)
````

## 5. Simple Login System (Maximum 3 Attempts)

**Problem Statement:**
Create a simple login system with a maximum of 3 password attempts using a loop.

````python
# Simulated function for demonstration, and interactive version below
def login_system_demo(simulated_inputs=None):
    correct_password = "admin@2026"
    max_attempts = 3
    attempts = 0
    
    while attempts < max_attempts:
        attempts += 1
        remaining = max_attempts - attempts
        
        if simulated_inputs is not None:
            pwd = simulated_inputs[attempts - 1]
            print(f"Attempt {attempts}/{max_attempts}: Entered '{pwd}'")
        else:
            pwd = input(f"Enter password ({remaining + 1} attempt(s) remaining): ")
            
        if pwd == correct_password:
            print("Access Granted! Welcome to the system.")
            break
        else:
            if remaining > 0:
                print(f"Incorrect password! {remaining} attempt(s) left.")
            else:
                print("Account Locked! Maximum attempts exceeded.")

print("--- Login Demonstration ---")
login_system_demo(simulated_inputs=["wrong123", "testpass", "admin@2026"])

# To try interactively, uncomment below:
# login_system_demo()
````

## 7. Accept 10 Numbers from the User into a List

**Problem Statement:**
Accept 10 numbers from the user and store them in a list using a loop.

````python
def collect_ten_numbers(simulated_data=None):
    numbers_list = []
    total_count = 10
    
    for i in range(1, total_count + 1):
        if simulated_data is not None:
            val = simulated_data[i - 1]
            print(f"Input {i}/{total_count}: {val}")
            numbers_list.append(val)
        else:
            while True:
                try:
                    val = float(input(f"Enter number {i}/{total_count}: "))
                    numbers_list.append(val)
                    break
                except ValueError:
                    print("Invalid input! Please enter a valid number.")
                    
    return numbers_list

print("--- Demonstration with 10 sample inputs ---")
demo_inputs = [12, 45, 78, 23, 56, 89, 90, 11, 34, 67]
result = collect_ten_numbers(simulated_data=demo_inputs)

print(f"\nCollected Numbers List: {result}")
print(f"Total Elements Stored : {len(result)}")

# To try interactively, uncomment below:
# interactive_list = collect_ten_numbers()
# print("User List:", interactive_list)
````

## 8. Character Frequency Counter (without Counter)

**Problem Statement:**
Count the frequency of every character in a string **without using `collections.Counter`**.

**Example:**
```
banana
b -> 1
a -> 3
n -> 2
```

````python
def count_character_frequency(text):
    frequency_dict = {}
    
    # Iterate through characters and count occurrences
    for char in text:
        if char in frequency_dict:
            frequency_dict[char] += 1
        else:
            frequency_dict[char] = 1
            
    return frequency_dict

# Test with 'banana'
word1 = "banana"
freq1 = count_character_frequency(word1)
print(f"Input string: '{word1}'")
print("Character frequencies:")
for char, count in freq1.items():
    print(f"{char} -> {count}")

print()

# Additional test with 'programming'
word2 = "programming"
freq2 = count_character_frequency(word2)
print(f"Input string: '{word2}'")
print("Character frequencies:")
for char, count in freq2.items():
    print(f"{char} -> {count}")
````

## 9. Find Second-Largest Number (without sort)

**Problem Statement:**
Find the second-largest number in a list **without using `sort()`**.

````python
def find_second_largest(numbers):
    if len(numbers) < 2:
        return None, None
    
    largest = float('-inf')
    second_largest = float('-inf')
    
    for num in numbers:
        if num > largest:
            second_largest = largest
            largest = num
        elif num > second_largest and num < largest:
            # Update second_largest only if it's strictly less than largest (handles duplicates)
            second_largest = num
            
    return largest, second_largest

# Test Case 1: Positive list with duplicate maximums
nums1 = [45, 12, 85, 32, 89, 39, 69, 89, 23]
max1, sec1 = find_second_largest(nums1)
print(f"Input List: {nums1}")
print(f"Largest Number       : {max1}")
print(f"Second-Largest Number: {sec1}")

print()

# Test Case 2: Negative numbers list
nums2 = [-10, -50, -5, -20, -5, -12]
max2, sec2 = find_second_largest(nums2)
print(f"Testing another list: {nums2}")
print(f"Largest Number       : {max2}")
print(f"Second-Largest Number: {sec2}")
````

## 10. Check Palindrome Using Loops

**Problem Statement:**
Check whether a string is a palindrome using loops.

````python
def is_palindrome_loop(text):
    # Two-pointer loop comparison
    cleaned = str(text).lower().strip()
    n = len(cleaned)
    
    for i in range(n // 2):
        if cleaned[i] != cleaned[n - 1 - i]:
            return False
    return True

# Test cases
test_words = ["racecar", "madam", "banana", "radar", "hello", "level", "12321", "Python"]

print("=" * 50)
print("               PALINDROME TEST RESULTS            ")
print("=" * 50)
for word in test_words:
    status = " IS a Palindrome" if is_palindrome_loop(word) else "NOT a Palindrome"
    print(f"• {repr(word):<20} ->  {status}")
print("=" * 50)
````

## 12. Basic ATM Simulation

**Problem Statement:**
Create a basic ATM simulation where a user can repeatedly:
- **Check balance**
- **Deposit money**
- **Withdraw money**
- **Exit**

````python
def run_atm(simulated_actions=None):
    balance = 10000.00
    print("=" * 40)
    print("       WELCOME TO PYTHON BANK ATM       ")
    print("=" * 40)
    
    # If simulated actions are passed, run automated demo
    if simulated_actions is not None:
        for action in simulated_actions:
            op = action.get("choice")
            if op == "1":
                print("\n[Simulation Step]: Selected Option 1 (Check Balance)")
                print(f"-> Current Balance: ₹{balance:,.2f}")
            elif op == "2":
                amt = action.get("amount", 0)
                print(f"\n[Simulation Step]: Selected Option 2 (Deposit ₹{amt:,.2f})")
                if amt > 0:
                    balance += amt
                    print(f"-> Successfully deposited ₹{amt:,.2f}.")
                    print(f"-> Updated Balance: ₹{balance:,.2f}")
                else:
                    print("-> Error: Deposit amount must be positive.")
            elif op == "3":
                amt = action.get("amount", 0)
                label = action.get("note", f"Withdraw ₹{amt:,.2f}")
                print(f"\n[Simulation Step]: Selected Option 3 ({label})")
                if amt <= 0:
                    print("-> Error: Withdrawal amount must be positive.")
                elif amt > balance:
                    print(f"-> Error: Insufficient funds! Available balance: ₹{balance:,.2f}")
                else:
                    balance -= amt
                    print(f"-> Successfully withdrew ₹{amt:,.2f}.")
                    print(f"-> Updated Balance: ₹{balance:,.2f}")
            elif op == "4":
                print("\n[Simulation Step]: Selected Option 4 (Exit)")
                print("-> Thank you for banking with us! Have a great day.")
                break
        return

    # Interactive loop
    while True:
        print("\n--- ATM Menu ---")
        print("1. Check Balance")
        print("2. Deposit Money")
        print("3. Withdraw Money")
        print("4. Exit")
        
        choice = input("Enter your choice (1-4): ").strip()
        
        if choice == '1':
            print(f"\n[Balance]: Your current balance is ₹{balance:,.2f}")
            
        elif choice == '2':
            try:
                amount = float(input("Enter deposit amount: ₹"))
                if amount > 0:
                    balance += amount
                    print(f"Successfully deposited ₹{amount:,.2f}.")
                    print(f"Updated Balance: ₹{balance:,.2f}")
                else:
                    print("Error: Deposit amount must be greater than 0.")
            except ValueError:
                print("Error: Invalid amount entered.")
                
        elif choice == '3':
            try:
                amount = float(input("Enter withdrawal amount: ₹"))
                if amount <= 0:
                    print("Error: Withdrawal amount must be greater than 0.")
                elif amount > balance:
                    print(f"Error: Insufficient funds! Current balance is ₹{balance:,.2f}")
                else:
                    balance -= amount
                    print(f"Successfully withdrew ₹{amount:,.2f}.")
                    print(f"Updated Balance: ₹{balance:,.2f}")
            except ValueError:
                print("Error: Invalid amount entered.")
                
        elif choice == '4':
            print("\nThank you for banking with us. Have a wonderful day!")
            break
        else:
            print("Invalid choice! Please select an option between 1 and 4.")

# Run simulation demo
demo_workflow = [
    {"choice": "1"},
    {"choice": "2", "amount": 2500.00},
    {"choice": "3", "amount": 4000.00},
    {"choice": "3", "amount": 15000.00, "note": "Attempt Overdraft ₹15,000.00"},
    {"choice": "4"}
]
run_atm(simulated_actions=demo_workflow)

# To launch the interactive ATM prompt, uncomment:
# run_atm()
````
